```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv(override=True)
google_api_key = os.getenv('GOOGLE_API_KEY')

gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini_model = 'gemini-3.5-flash-lite'


gemini = OpenAI(api_key=google_api_key, base_url=gemini_url)
```

```python
system_message_test = "Você é um assistente muito prestativo."

def message_gemini(prompt):
    messages = [
        {"role": "system", "content": system_message_test},
        {"role": "user", "content": prompt}
    ]
    response = gemini.chat.completions(model=gemini_model, messages=messages)
    return response.choinces[0].message.content
```

```python
message_gemini("Que dia é hoje?")
```
